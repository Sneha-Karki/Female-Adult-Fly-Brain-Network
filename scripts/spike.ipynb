```python
import pandas as pd
import json

# Load the spike data
mapping = pd.read_csv("../Data/spike_data/spike_data/id_to_idx.csv")

with open("../Data/spike_data/spike_data/spike_data.json") as f:
    spikes = json.load(f)

# Use your existing circuit neurons
jo_nodes = pd.read_parquet("../Data/jo circuit/jo_nodes.parquet")

# Map FlyWire root ID -> simulation index
root_to_idx = dict(zip(
    mapping.iloc[:, 0].astype(str),
    mapping.iloc[:, 1].astype(int)
))

# Find circuit neurons present in simulation
circuit_ids = set(jo_nodes["id"].astype(str))

circuit_indices = {
    root_to_idx[root_id]
    for root_id in circuit_ids
    if root_id in root_to_idx
}

# Find circuit neurons that actually fired
fired_indices = set()
circuit_spike_count = 0

for fired_neurons in spikes.values():
    for idx in fired_neurons:
        if idx in circuit_indices:
            fired_indices.add(idx)
            circuit_spike_count += 1

print("Circuit neurons:", len(circuit_ids))
print("Circuit neurons mapped to simulation:", len(circuit_indices))
print("Circuit neurons that fired:", len(fired_indices))
print("Total circuit spike events:", circuit_spike_count)

print("Time steps with events:", len(spikes))
print("Last time step:", max(map(int, spikes.keys())))
```

```python
# Find the 3 circuit neurons missing from the simulation
missing = jo_nodes[
    ~jo_nodes["id"].astype(str).isin(root_to_idx)
]

print("Missing neurons:", len(missing))

display(missing)
```
